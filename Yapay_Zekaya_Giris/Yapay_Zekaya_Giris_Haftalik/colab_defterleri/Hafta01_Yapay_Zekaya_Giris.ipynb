## 2. NumPy ile vektör ve matris işlemleri (Çözümlü Örnek 1.4)

```python
import numpy as np

x = np.array([2, 1, 3])            # girdi vektörü
w = np.array([0.5, -1.0, 0.2])     # ağırlıklar
b = 0.4                            # bias

z = np.dot(w, x) + b               # iç çarpım + bias
W = np.array([[1, 0, 2],
              [-1, 3, 1]])

print(round(z, 3))                 # 1.0
print(np.linalg.norm(x))           # 3.7416573867739413
print(W @ x)                       # [8 4]
```

## 3. A* araması (Çözümlü Örnek 1.3)
Haritadaki yol maliyetleri `graf`, sezgisel tahminler `h` sözlüğünde. Algoritma her adımda f = g + h değeri en küçük düğümü genişletir.

```python
import heapq
# yol maliyetleri (S'den G'ye doğru) ve hedefe kuş uçuşu tahminler
graf = {"S": {"A": 1, "B": 4}, "A": {"B": 2, "C": 5},
        "B": {"C": 2, "G": 6}, "C": {"G": 3}, "G": {}}
h = {"S": 5, "A": 4, "B": 2, "C": 3, "G": 0}

def a_yildiz(bas, hedef):
    acik = [(h[bas], 0, bas, [bas])]        # (f, g, düğüm, yol)
    en_iyi = {bas: 0}
    while acik:
        f, g, n, yol = heapq.heappop(acik)  # f'si en küçük düğüm
        if g > en_iyi[n]: continue          # eskimiş kayıt
        print(f"genişlet {n}: g={g}, h={h[n]}, f={f}")
        if n == hedef: return yol, g
        for m, c in graf[n].items():
            if g + c < en_iyi.get(m, float("inf")):
                en_iyi[m] = g + c
                heapq.heappush(acik, (g+c+h[m], g+c, m, yol+[m]))
yol, maliyet = a_yildiz("S", "G")
print(" → ".join(yol), "| maliyet:", maliyet)
```

### Deneyin: h = 0 alınırsa A* tek biçimli maliyet aramasına (UCS) dönüşür

```python
h = {d: 0 for d in h}        # sezgisel bilgi yok
yol, maliyet = a_yildiz("S", "G")
print(" → ".join(yol), "| maliyet:", maliyet)
```
